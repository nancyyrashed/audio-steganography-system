### Task 3.3: Investigation into a Modern Audio Steganography Method

For this task, I decided to research a steganography method not covered in the module and settled on **Echo Hiding**. It caught my attention because it uses a completely different approach from what we used in Task 3.2 with LSB (Least Significant Bit) methods. Where LSB involves directly tweaking digital bits of an audio file, echo hiding relies on inserting short, carefully placed echoes into the signal.

---

#### How It Works

Echo hiding embeds data by introducing a faint echo into the audio, delayed by a small amount — usually 1 to 2 milliseconds. These delays are so short that human ears can't detect them. A specific delay value represents a binary “0” and another represents a “1”. So, for example, a 1ms delay might mean “0” and a 2ms delay might mean “1”. You take the original audio, duplicate it, delay it slightly, and mix it back in.

To read the hidden data, signal analysis is used to detect those small echo delays. This usually involves autocorrelation or cepstrum analysis — basically tools that can pick out repetitive patterns or delays inside the waveform.

---

#### Technical Details

This method doesn’t mess with bits directly. Instead, it uses digital signal processing (DSP) tools to work with the waveform. You need some DSP knowledge to implement this properly, especially to manage echo parameters like delay time and amplitude. For decoding, it requires a clean signal and careful tuning to detect the echoes, especially if there's noise or music complexity.

---

#### Strengths

One major benefit is how well it hides the message without affecting audio quality. Since it doesn’t alter samples directly, the result sounds exactly like the original. It’s also more robust than LSB — it can survive minor processing like resampling or filtering. This makes it a better option when you want to hide something in an audio file without making it obvious or risking accidental removal during playback or editing.

---

#### Weaknesses

However, echo hiding has some downsides. It’s harder to implement and needs accurate echo placement. It's also less effective with audio that already has natural reverb or echo — the embedded data could get lost in the background. Plus, the data rate is low, so it's only good for short messages or codes.

---

#### Comparison to LSB

Compared to the LSB method, echo hiding is much stealthier but not as simple or as flexible. LSB is fast and easy but fragile — one audio conversion can ruin the hidden message. Echo hiding is slower but holds up better under typical audio changes.

---

#### Final Thoughts

Overall, echo hiding is a more advanced but practical option when you care more about security and stealth than speed. It’s better suited for sensitive use cases where detection must be avoided — definitely more in line with covert communication than general use.
